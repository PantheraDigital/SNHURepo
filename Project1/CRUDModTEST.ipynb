```python
import CRUDMod

shelter_db = CRUDMod.AnimalShelter()

print("CREATE TEST")
tests = [{"dog":""}, {"dog":"dog test data"}, {}, ["dog","dog test data"]]
for test in tests:
    print("---" + str(test))
    result = shelter_db.create(test)
    if result == True:
        print("pass")
    else:
        print("fail")
```

```python
import CRUDMod

shelter_db = CRUDMod.AnimalShelter()

print("READ TEST")
tests = [{"dog":""}, {"dog":"dog test data"}, {}, "dog"]
for test in tests:
    print("---" + str(test))
    result = shelter_db.read(test)
    if isinstance(result, list):
        if len(result) > 0:
            print("pass")
            continue
    print("fail")
```

```python
import CRUDMod
shelter_db = CRUDMod.AnimalShelter()

print("Empty Test")
print(shelter_db.read({"cat":"example data"}))
result = shelter_db.update({"cat":"example data"}, {"$set": {"cat": "new data"}})
print("modified: ",result)
print("result: ",shelter_db.read({"cat":"example data"}))

print("\nSingle Update Test")
shelter_db.create({"cat":"example data"})
print(shelter_db.read({"cat":"example data"}))
result = shelter_db.update({"cat":"example data"}, {"$set": {"cat": "new data"}})
print("modified: ",result)
print("result: ",shelter_db.read({"cat":"new data"}))

print("\nMulti Update Test")
shelter_db.create({"cat":"new data"})
print(shelter_db.read({"cat":"new data"}))
result = shelter_db.update({"cat":"new data"}, {"$set": {"cat": "old data"}})
print("modified: ",result)
print("result: ",shelter_db.read({"cat":"old data"}))
```

```python
import CRUDMod
shelter_db = CRUDMod.AnimalShelter()

print("Empty Test")
print(shelter_db.read({"cat":"example data"}))
result = shelter_db.delete({"cat":"example data"})
print("modified: ",result)
print("result: ",shelter_db.read({"cat":"example data"}))

print("\nSingle Update Test")
shelter_db.create({"cat":"example data"})
print(shelter_db.read({"cat":"example data"}))
result = shelter_db.delete({"cat":"example data"})
print("modified: ",result)
print("result: ",shelter_db.read({"cat":"new data"}))

print("\nMulti Update Test")
shelter_db.create({"cat":"new data"})
shelter_db.create({"cat":"new data"})
print(shelter_db.read({"cat":"new data"}))
result = shelter_db.delete({"cat":"new data"})
print("modified: ",result)
print("result: ",shelter_db.read({"cat":"old data"}))
```
